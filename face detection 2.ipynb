```python
import cv2
import os
import numpy as np

image_folder = r"C:\Users\Administrator\Downloads\face detection project\images"
output_folder = r"C:\Users\Administrator\Downloads\face detection project\detected_faces"
model_path = r"C:\Users\Administrator\Downloads\face detection project\lbfmodel.yaml"

os.makedirs(output_folder, exist_ok=True)

face_detector = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)
if face_detector.empty():
    raise Exception("Cannot load face detector.")

facemark = cv2.face.createFacemarkLBF()
facemark.loadModel(model_path)

face_index = 0
LEFT_EYE = list(range(36, 42))
RIGHT_EYE = list(range(42, 48))
NOSE = list(range(27, 36))

def crop_region(image, points, padding=10):
    pts = np.array(points, dtype=np.int32)
    x, y, w, h = cv2.boundingRect(pts)
    x = max(0, x - padding)
    y = max(0, y - padding)
    w = min(image.shape[1] - x, w + padding * 2)
    h = min(image.shape[0] - y, h + padding * 2)
    return image[y:y+h, x:x+w]

for image_name in os.listdir(image_folder):
    image_path = os.path.join(image_folder, image_name)
    image = cv2.imread(image_path)
    if image is None:
        continue

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    faces = face_detector.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(40, 40)
    )

    if len(faces) == 0:
        print(f"{image_name}: No faces detected.")
        continue

    success, landmarks = facemark.fit(gray, faces)

    if not success:
        print(f"{image_name}: Landmarks not detected.")
        continue

    for face_landmarks in landmarks:
        points = face_landmarks.reshape(-1, 2)

        left_eye_points = [points[i] for i in LEFT_EYE]
        right_eye_points = [points[i] for i in RIGHT_EYE]
        nose_points = [points[i] for i in NOSE]

        face_folder = os.path.join(output_folder, f"person_{face_index}")
        os.makedirs(face_folder, exist_ok=True)

        x, y, w, h = faces[face_index] if face_index < len(faces) else faces[0]
        face = image[y:y+h, x:x+w]
        cv2.imwrite(os.path.join(face_folder, "face.jpg"), face)

        left_eye_crop = crop_region(image, left_eye_points)
        cv2.imwrite(os.path.join(face_folder, "left_eye.jpg"), left_eye_crop)

        right_eye_crop = crop_region(image, right_eye_points)
        cv2.imwrite(os.path.join(face_folder, "right_eye.jpg"), right_eye_crop)

        nose_crop = crop_region(image, nose_points)
        cv2.imwrite(os.path.join(face_folder, "nose.jpg"), nose_crop)

        face_index += 1

print("Done.")
```
